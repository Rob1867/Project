# Define Functions

Define a function to Find the first row containing 'search_text', remove rows above it, and set it as column names.

```python
def clean_dataframe(df, search_text):

    # Find the index of the first occurrence of the search text
    idx = df[df.apply(lambda row: row.astype(str).str.contains(search_text, case=False).any(), axis=1)].index
    
    if idx.empty:
        raise ValueError(f"'{search_text}' not found in the DataFrame.")
    
    # Get the first matching index
    first_match_idx = idx[0]
    
    # Update the DataFrame by removing rows above the match
    df = df.iloc[first_match_idx:].reset_index(drop=True)
    
    # Set the first row as column names and drop it from the DataFrame
    df.columns = df.iloc[0]
    df = df[1:].reset_index(drop=True)
    
    return df
```

Define a function to check if Script Activity IDs all have unique date

```python
def check_consistent_dates(df, reference_col, date_col):
    """
    Checks if a given reference number always corresponds to the same date
    in a Pandas DataFrame.

    Args:
        df (pd.DataFrame): The input DataFrame.
        reference_col (str): The name of the column containing reference numbers.
        date_col (str): The name of the column containing dates.

    Returns:
        dict: A dictionary indicating consistency.
              - If all consistent: {'consistent': True}
              - If inconsistent: {'consistent': False, 'violating_references': list_of_references}
    """
    # Group by the reference column and count the number of unique dates for each reference
    date_counts = df.groupby(reference_col)[date_col].nunique()

    # Identify reference numbers that have more than one unique date
    violating_references = date_counts[date_counts > 1].index.tolist()

    if not violating_references:
        return {'consistent': True}
    else:
        return {'consistent': False, 'violating_references': violating_references}
```

Define a function to return the maximum script version for each script in the visual condition data, for comparision with those in df_Script_Versions. If script version in data is higher, this will need checking to ensure there is still an importance score for all questions

```python
def get_max_script_versions(df):
    # Get unique Script IDs
    unique_script_ids = df['Script ID'].unique()
    
    # Initialize a dictionary to store results
    max_versions = {}
    
    # Loop through each unique Script ID
    for script_id in unique_script_ids:
        # Filter the DataFrame for the current Script ID
        filtered_df = df[df['Script ID'] == script_id]
        
        # Find the maximum Script Version Number
        max_version = filtered_df['Script Version Number'].max()
        
        # Store the result in the dictionary
        max_versions[script_id] = max_version
    
    # Convert the dictionary to a DataFrame
    result_df = pd.DataFrame(list(max_versions.items()), columns=['Script ID', 'Script Version Number'])
    
    return result_df
```

```python
# Example setup
# target_df has headers only
target_df = pd.DataFrame(columns=["Asset_Name", "Location", "Class", "WCS", "Criticality"])

# source_df has data (with different column names)
# e.g., source columns: ["Name", "Site", "Category", "WCS", "RiskScore", ...]
# source_df = ...

# 1) Filter rows in source where WCS >= 0.75
filtered_src = df_unique_clean.loc[df_unique_clean["WCS"] >= 0.75].copy()

# 2) Map columns: source -> target
column_map = {
    "Script Activity ID": "Unique ID",
    "Plant Number": "Functional Position",
    "Script Description": "How was the Defect Discovered?"
}

# 3) Keep only columns that exist in source AND target
valid_map = {src: tgt for src, tgt in column_map.items()
             if src in filtered_src.columns and tgt in df_etip.columns}

# 4) Reindex target_df to match filtered source index (so assignment aligns)
target_df = target_df.reindex(filtered_src.index)

# 5) Assign across
for src_col, tgt_col in valid_map.items():
    df_etip[tgt_col] = filtered_src[src_col]

# target_df now contains rows where WCS >= 0.75 with the mapped columns filled
```

```python
# --- Inputs: adjust these names to your actual columns ---
links_df = df_vi_civil                    # dataframe containing Civil Item and links
etip_df = df_etip                       # your target dataframe
civil_col = "Civil Item"                # column that contains the text 'civil photo'
links_key_col = "Script Activity ID"    # key in df_links
etip_key_col = "Unique ID"              # key in df_etip
link_col = "Document Link"                  # column that contains the actual URLs
output_col = "Links to Pictures (If available)"  # target column in df_etip
sep = " | "                             # separator for concatenated links

# --- 0) Defensive checks ---
for dfname, dfobj, cols in [
    ("df_vi_civil", links_df, [civil_col, links_key_col, link_col]),
    ("df_etip", etip_df, [etip_key_col]),
]:
    missing = [c for c in cols if c not in dfobj.columns]
    if missing:
        raise KeyError(f"{dfname} missing required columns: {missing}")

# Ensure output column exists (optional; pandas will create it on assignment anyway)
if output_col not in etip_df.columns:
    etip_df[output_col] = pd.NA

# --- 1) Filter rows that represent civil photos (case-insensitive) ---
mask_civil_photo = links_df[civil_col].astype(str).str.contains("civil photo", case=False, na=False)
civil_links = links_df.loc[mask_civil_photo, [links_key_col, link_col]].copy()

# Clean links: strip whitespace, drop empty
civil_links[link_col] = civil_links[link_col].astype(str).str.strip()
civil_links = civil_links[civil_links[link_col].ne("")]

# --- 2) Deduplicate links per key ---
civil_links = civil_links.drop_duplicates(subset=[links_key_col, link_col])

# --- 3) Concatenate links per Script Activity ID ---
links_per_id = (
    civil_links
    .groupby(links_key_col, as_index=False)[link_col]
    .agg(lambda s: sep.join(s))
    .rename(columns={link_col: output_col})
)

lookup = links_per_id.set_index(links_key_col)[output_col]

etip_df[output_col] = etip_df[etip_key_col].map(lookup)
```
